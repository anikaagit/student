## Part 1 — What `new` actually does

You've typed `new Scanner(...)` since Lesson 1.4. Here's what that word was doing every single time, broken into three steps:

```
   Scanner order = new Scanner("Taro 3");
                    ^^^ ^^^^^^^^^^^^^^^^
                     |         |
                     |         └─ 2. CONSTRUCT: build the object, set it up
                     └─ 1. ALLOCATE: reserve a spot in memory for it

                                          3. RETURN a reference to that spot,
                                             which gets stored in "order"
```

1. **Allocate** — Java reserves a block of memory (a region called the **heap**) big enough to hold the new object.
2. **Construct** — a special block of code called a **constructor** runs, setting the object up.
3. **Return a reference** — the address of that memory gets handed back, and that's what your variable actually stores.

This is the picture from Lesson 1.12 made precise: the variable `order` was never the `Scanner` itself. It's the address `new` returned in step 3.

````java
import java.util.Scanner;

public class NewInThreeSteps {
    public static void main(String[] args) {
        // Step 1: allocate space on the heap for a new Scanner
        // Step 2: run the Scanner constructor, giving it "Taro 3" to work with
        // Step 3: return a reference, stored in the variable "order"
        Scanner order = new Scanner("Taro 3");

        System.out.println("First token: " + order.next());
        System.out.println("Second token: " + order.next());
    }
}

NewInThreeSteps.main(null);
````

## Part 2 — What a constructor actually is

> **A constructor is a special block of code that runs automatically when an object is created with `new`. Its job is to set the object up.**

A few things make a constructor different from every method you've written since Lesson 1.9:

| Ordinary method | Constructor |
|---|---|
| Has a return type (`void`, `int`, `double`, ...) | **Has no return type at all** — not even `void` |
| Called explicitly, whenever you want | Called automatically, **only** by `new` |
| Any name you choose | **Must match the class name exactly** |

You've never written a constructor yourself — that's Unit 3, once you start designing your own classes — but you've been *calling* one every time you wrote `new Scanner(...)` or `new String(...)`. The parentheses after `Scanner` or `String` are exactly the same kind of argument list you learned to match to a signature back in Lesson 1.9. Different arguments can call a different constructor, the same way different arguments call a different overload.

````java
public class ConstructorArguments {
    public static void main(String[] args) {
        // Two DIFFERENT constructors of the String class, chosen by argument type --
        // exactly like choosing an overload in Lesson 1.9.
        String fromLiteral = new String("Taro Milk Tea");         // String(String)
        char[] letters = {'B', 'o', 'b', 'a'};
        String fromChars = new String(letters);                   // String(char[])

        System.out.println(fromLiteral);
        System.out.println(fromChars);
    }
}

ConstructorArguments.main(null);
````

Both lines built a real `String` object, using two entirely different constructors on the same class — one takes a `String`, the other takes an array of characters. The constructor decided how the object's initial state got set up; you never had to know how it did that internally, which is abstraction again from Lesson 1.7.

## Part 3 — Declaring, constructing, and assigning together

`Scanner order = new Scanner("Taro 3");` is doing three things from earlier lessons **in a single line**:

| Piece | What it is | From Lesson |
|---|---|---|
| `Scanner order` | **Declaration** — creates the variable, gives it a type | 1.4 |
| `new Scanner("Taro 3")` | Allocates and constructs the object | This lesson |
| `=` | **Assignment** — stores the result in `order` | 1.4 |

Nothing new is happening syntactically that you haven't already seen — declaring and initializing on one line worked exactly this way for primitives back in Lesson 1.4. The only difference is that the right-hand side, instead of being a literal like `5.75`, is a full construction: allocate, build, and hand back a reference.

You can still split it across two lines, exactly as you could for a primitive:

````java
import java.util.Scanner;

public class SplitDeclaration {
    public static void main(String[] args) {
        Scanner order;                          // DECLARATION only
        order = new Scanner("Matcha Latte 2");  // construction + ASSIGNMENT, later

        System.out.println(order.next());
    }
}

SplitDeclaration.main(null);
````

## Part 4 — Aliasing: two names, one object

This is the manager's duplicate-ticket problem, and it's the core idea of this lesson.

> **Aliasing** happens when two or more reference variables hold a reference to the **same** object. There is still only one object — just multiple names that can reach it.

Compare this directly to Lesson 1.12, Part 2, where two `Scanner` objects were completely **independent** — separate `new` calls, separate memory, separate everything. This time, watch what happens when a second variable is assigned from the **first variable itself**, instead of from a fresh `new` call.

````java
import java.util.Scanner;

public class Aliasing {
    public static void main(String[] args) {
        Scanner original = new Scanner("Taro Milk Tea");
        Scanner alias = original;     // NOT "new Scanner(...)" -- just a second name for the SAME object

        System.out.println("original reads: " + original.next());   // "Taro"
        System.out.println("alias reads next: " + alias.next());    // "Milk" -- NOT "Taro" again!
    }
}

Aliasing.main(null);
````

`alias` picked up **exactly where `original` left off** — `Milk`, not `Taro` again. If these were two separate objects (like `scanA` and `scanB` back in Lesson 1.12), `alias` would have started fresh at `Taro`. Instead, `alias.next()` advanced the *one and only* `Scanner` that both names point to.

```
                object still exists once
  original ────────┐
                    ├──────►  [Scanner object: "Taro Milk Tea", position: 1]
  alias ────────────┘
```

This is precisely the duplicate-ticket scenario: `original` and `alias` are two different tickets with two different names printed on them, but they both point at the same physical order. Mark it fulfilled through one name, and it's fulfilled through the other too — because there's only one order to begin with.

## Part 5 — `==` on reference types: identity, not content

You've used `==` on primitives since Lesson 1.2, where it compares values directly: `5 == 5` is `true` because both sides really are the number 5.

`==` on a reference type asks a **completely different question**:

> **For reference types, `==` checks whether two variables point to the exact same object in memory — not whether the objects contain equal-looking data.**

Two objects can hold identical content and still fail an `==` check, because `==` never looks inside the object at all — it only compares the addresses.

````java
import java.util.Scanner;

public class IdentityCheck {
    public static void main(String[] args) {
        Scanner a = new Scanner("Taro");
        Scanner b = a;                     // ALIAS -- same object as a
        Scanner c = new Scanner("Taro");   // a SEPARATE object, even with equal starting content

        System.out.println("a == b (alias, same object)?          " + (a == b));
        System.out.println("a == c (separate objects, same data)? " + (a == c));
    }
}

IdentityCheck.main(null);
````

`a == b` is `true` — they're two names for one object, exactly like Part 4. `a == c` is `false` — even though `c` started with the identical input string `"Taro"`, it's a **different object** sitting at a different address. `==` doesn't care that they'd behave identically if you used them the same way; it only cares whether they're literally the same object.

> **A warning worth remembering now:** `String` has a quirk that makes `==` behave unpredictably on it — sometimes two separately-written string literals *do* pass an `==` check, for reasons that have nothing to do with what you're testing for. Comparing `String` content correctly is the actual subject of Topic **1.15**, and the safe method there is `.equals(...)`, not `==`. For any reference type, though, the rule you just saw holds: `==` always means "same object," never "equal content."

## Part 6 — Passing a reference to a method

Back in Lesson 1.9 and 1.10 you passed primitives into methods. What happens when the parameter is a **reference type**?

> **A method parameter of a reference type receives a copy of the reference — but that copy still points to the exact same object the caller has.**

The method doesn't get its own independent object. It gets another alias, valid only for the life of that method call — but still pointing at the one real object.

````java
import java.util.Scanner;

public class PassingAReference {

    public static void consumeOne(Scanner s) {
        // "s" is a separate variable from "order" in main, but it points to
        // the SAME Scanner object. Advancing it here really does advance the original.
        System.out.println("  inside consumeOne, reading: " + s.next());
    }

    public static void main(String[] args) {
        Scanner order = new Scanner("Taro Milk Tea");

        consumeOne(order);   // pass the reference in

        // The token consumed INSIDE the method is genuinely gone now.
        System.out.println("back in main, next token is: " + order.next());
    }
}

PassingAReference.main(null);
````

`order` never got passed "into" the method as a fresh copy of a `Scanner`. `consumeOne` received a reference to the **exact same object**, called it `s`, and consumed a token through that name. By the time control returned to `main` (the detour idea from Lesson 1.10), `order`'s own position had genuinely moved forward — because `s` and `order` were aliases for one object the entire time.

This is a real, important contrast with how primitives behave as parameters: a method **cannot** change a primitive argument's value back in the caller (you'll see that directly in Unit 3), but a method genuinely **can** change the state of an object it receives, because it's holding a reference to the real thing, not a disposable copy of a value.

## Hack 2 — Alias or independent?

For each pair, predict whether they're **aliases** (same object) or **independent** (separate objects), and predict what the second `.next()` call in each pair returns. Fill in your predictions before running.


| Pair | Alias or independent? | `p1.next()` / `q1.next()` result | Second call's result |
|---|---|---|---|
| `p1` / `p2` | | | |
| `q1` / `q2` | | | |

````java
import java.util.Scanner;

public class AliasOrIndependent {
    public static void main(String[] args) {
        Scanner p1 = new Scanner("A B C");
        Scanner p2 = p1;

        Scanner q1 = new Scanner("X Y Z");
        Scanner q2 = new Scanner("X Y Z");

        System.out.println("p1.next(): " + p1.next());
        System.out.println("p2.next(): " + p2.next());

        System.out.println("q1.next(): " + q1.next());
        System.out.println("q2.next(): " + q2.next());
    }
}

AliasOrIndependent.main(null);
````

<details>
<summary><b>Check your answers</b></summary>

`q1` and `q2` are **independent** — each was built with its own `new Scanner(...)` call, even though they started with identical text. `q1.next()` returns `X`, and `q2` — a completely separate object — also starts fresh at its own beginning and returns `X` too, not `Y`.

The surface-level content was the same for both pairs (`"A B C"`-style input), but the *relationship* between the variables was completely different, and that relationship — not the content — is what determined the output.
</details>

## Hack 3 — Fix the double-fulfilled ticket

The cafe's ticket system has a bug matching the manager's story from the top of this lesson. Two runners are supposed to work on **two separate orders**, but the code accidentally aliases them. Run it, find the bug, then fix it so each runner truly gets an independent order.

Correct behavior: `runner1` should read `Taro`, then `Milk`. `runner2` should independently read `Matcha`, then `Latte`.

````java
import java.util.Scanner;

public class DuplicateTicket {
    public static void main(String[] args) {
        Scanner runner1 = new Scanner("Taro Milk");
        Scanner runner2 = runner1;    // BUG: this doesn't create a second order

        System.out.println("runner1 sees: " + runner1.next());
        System.out.println("runner2 sees: " + runner2.next());   // should be "Matcha", isn't!
    }
}

DuplicateTicket.main(null);
````

**Your bug report** (double-click to edit):

What's wrong, and what's your fix?

<details>
<summary><b>Check your answer</b></summary>

`runner2 = runner1;` makes `runner2` an **alias** for `runner1`'s object instead of a genuinely separate order. There was only ever one `Scanner`, reading `"Taro Milk"` — `runner2.next()` just continued reading from wherever `runner1` had already gotten to, which is why it returned `Milk` instead of a fresh `Matcha`.

Fix by giving `runner2` its own `new` call, building a real second object:


Now each runner has an independent order, exactly like `q1`/`q2` in Hack 2 — and each `.next()` call only affects the one object it's actually called on.

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Heap | The region of memory where objects created with `new` are stored |
| Constructor | A special block with no return type and the class's own name, run automatically by `new` |
| Aliasing | Two or more variables referencing the exact same object |
| `==` (on references) | Checks whether two variables point to the same object, not whether their contents match |

### The five things that will show up on the exam

1. `new` does three things: **allocate**, **construct**, **return a reference**.
2. A constructor has **no return type** and **shares the class's name**.
3. `variable2 = variable1;` (no `new`) creates an **alias**, not a new object.
4. `==` on reference types checks **identity**, never content.
5. A method parameter of a reference type is an **alias** for the caller's object — changes to the object's state are visible after the method returns.
