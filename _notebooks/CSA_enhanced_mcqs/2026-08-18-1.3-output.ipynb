## Part 1 — `print` vs `println`

You've been using `System.out.println` since Lesson 1. It has a sibling.

| Statement | What it does |
|---|---|
| `System.out.print(x)` | Displays `x` and **leaves the cursor where it is** |
| `System.out.println(x)` | Displays `x`, then **moves the cursor to a new line** |
| `System.out.println()` | Displays nothing, just moves to a new line — a blank line |

Think of the cursor as the tip of a marker on the cafe's chalkboard menu. `print` writes and leaves the marker where it stopped. `println` writes and then drops down to start a fresh line.

Run this and compare the two halves of the output.

````java
public class PrintVsPrintln {
    public static void main(String[] args) {
        // print -- everything lands on ONE line
        System.out.print("Taro");
        System.out.print("Matcha");
        System.out.print("Thai");

        System.out.println();          // empty println = just move to a new line

        // println -- each one starts a fresh line
        System.out.println("Taro");
        System.out.println("Matcha");
        System.out.println("Thai");
    }
}

PrintVsPrintln.main(null);
````

Notice `TaroMatchaThai` ran together with no spaces. `print` adds **nothing** — not a space, not a line break. If you want a space, you have to put one there yourself.

This matters on the exam. When a question shows a mix of `print` and `println` and asks what's displayed, you have to track the cursor line by line.

## Part 2 — String literals and escape sequences

> **Definition:** A **string literal** is fixed text written inside double quotation marks, like `"Taro Milk Tea"`.

Simple enough — until you need to print a character that has a special meaning in Java. How do you print a **double quote** when double quotes are what mark the beginning and end of your text?

That's what **escape sequences** are for. A backslash tells Java "the next character is data, not punctuation."

| Escape sequence | Prints | Use it for |
|---|---|---|
| `\"` | `"` | A quotation mark inside a string |
| `\\` | `\` | A single backslash (file paths, mostly) |
| `\n` | *new line* | Breaking one string across multiple lines |

Those three are the ones named in the CED. A fourth, `\t`, inserts a tab and is handy for lining up columns — you'll see it below.

The important mental shift: `\n` is **one character**, not two. The backslash isn't printed; it's an instruction to the compiler.

````java
public class MenuBoard {
    public static void main(String[] args) {
        System.out.println("Today's special: \"Brown Sugar Deluxe\"");   // \" prints a quote
        System.out.println("Recipe saved to C:\\cafe\\menu.txt");          // \\ prints one backslash
        System.out.println("Hot drinks\nCold drinks\nSmoothies");        // \n breaks the line

        System.out.println();

        System.out.println("Item\t\tPrice");                             // \t is a tab
        System.out.println("Taro\t\t$5.75");
        System.out.println("Matcha\t\t$6.00");
    }
}

MenuBoard.main(null);
````

Look at the second line of output carefully: you typed **four** backslashes in the source and got **two** in the display. Each `\\` in your code produces exactly one backslash on screen.

And notice the third statement was a *single* `println` that produced *three* lines of output. `\n` gives you line breaks without extra statements.

## Part 3 — Concatenation, and the trap hiding inside it

The `+` operator does two completely different jobs depending on what's on either side of it:

- **Number `+` number** → arithmetic. `5 + 3` gives `8`.
- **String `+` anything** → **concatenation**. The other value is converted to text and glued on.

`"Cups: " + 3` produces the text `Cups: 3`. The `3` stopped being a number the moment it touched a String.

````java
public class Concatenation {
    public static void main(String[] args) {
        String first  = "Brown Sugar";
        String second = "Milk Tea";

        System.out.println(first + " " + second);        // String + String
        System.out.println("Cups ordered: " + 3);        // String + int
        System.out.println("Price: $" + 5.75);           // String + double
        System.out.println("In stock? " + true);         // String + boolean
    }
}

Concatenation.main(null);
````

### The trap: `+` evaluates left to right

Here's the manager's `$122` bug. Java reads a chain of `+` operators **left to right**, one pair at a time. Once a String enters the chain, *everything after it* becomes concatenation.

Trace `"Total: " + 5 + 3` the way Java does:

```
  "Total: " + 5 + 3
   \_______ ______/
           v
       "Total: 5"      <- String + int gives a String
                 + 3
   \_____________ ___/
                 v
       "Total: 53"     <- String + int gives a String again
```

The `5` and the `3` never met each other. Now run it.

````java
public class ConcatTrap {
    public static void main(String[] args) {
        System.out.println("Total: " + 5 + 3);       // left to right: "Total: 5" then "Total: 53"
        System.out.println("Total: " + (5 + 3));     // parentheses force the math first
        System.out.println(5 + 3 + " drinks");       // math happens BEFORE the String appears
        System.out.println("Order " + 1 + 2 + 3);
    }
}

ConcatTrap.main(null);
````

Compare lines 1 and 3. Same operator, same values, opposite results — because in line 3 the two ints meet **before** the String joins the chain, so `+` is still doing arithmetic.

**The rule to memorize:** if you want math to happen inside a printed message, wrap it in parentheses. Every time.

```java
System.out.println("Total: " + (quantity * price));   // do this
```

## Part 4 — Arithmetic operators

Java gives you five arithmetic operators.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `4 + 3` | `7` |
| `-` | Subtraction | `4 - 3` | `1` |
| `*` | Multiplication | `4 * 3` | `12` |
| `/` | Division | `12 / 3` | `4` |
| `%` | Remainder ("mod") | `13 % 4` | `1` |

The first four you know from math class. `%` is probably new: it gives you **what's left over** after division. `13 % 4` is `1` because 4 goes into 13 three times with 1 left over.

`%` is far more useful than it looks. It answers questions like "is this number even?" (`n % 2 == 0`), "which leftovers didn't fit?", and "every 5th customer gets a free drink" (`n % 5 == 0`).

````java
public class BasicMath {
    public static void main(String[] args) {
        int scoopsPerCup = 3;
        int cups         = 4;

        System.out.println("Scoops needed:    " + (scoopsPerCup * cups));
        System.out.println("One more cup:     " + (cups + 1));
        System.out.println("One cancellation: " + (cups - 1));
        System.out.println("Even split of 12: " + (12 / 4));
        System.out.println("Leftover of 13:   " + (13 % 4));
    }
}

BasicMath.main(null);
````

## Part 5 — Integer division: the biggest trap in Unit 1

Read this rule twice, because the AP exam tests it constantly:

> **When both operands are `int`, the result is an `int`.** Java throws away the fractional part — it does **not** round.

So `7 / 2` is `3`. Not `3.5`, and not `4`. The `.5` is discarded, not rounded up.

This is where the manager's second bug came from: 25 ounces split between 2 cups was computed as `25 / 2`, which is `12`, and the missing half-ounce vanished silently.

### But it isn't a bug — it's a tool

Integer division and `%` are a matched pair. Together they answer "how many whole groups, and how many left over?"

The cafe has 20 cups to pack into trays that hold 6 each:

````java
public class TrayPacking {
    public static void main(String[] args) {
        int cups        = 20;
        int cupsPerTray = 6;

        int fullTrays = cups / cupsPerTray;    // how many COMPLETE trays
        int leftover  = cups % cupsPerTray;    // how many cups don't fill a tray

        System.out.println("Cups to pack:   " + cups);
        System.out.println("Full trays:     " + fullTrays);
        System.out.println("Cups left over: " + leftover);
        System.out.println("Check: " + fullTrays + " x " + cupsPerTray + " + " + leftover + " = " + cups);
    }
}

TrayPacking.main(null);
````

### Rescuing the fraction: mix in a `double`

The other half of the rule:

> **If at least one operand is a `double`, the result is a `double`.**

So you only need *one* side of the division to be a `double` to keep the decimal. Writing `2.0` instead of `2` is enough.

````java
public class SplitTheDrink {
    public static void main(String[] args) {
        int totalOunces = 25;
        int cups        = 2;

        System.out.println("int / int:        " + (totalOunces / cups));    // 12  -- the .5 is gone
        System.out.println("int / double:     " + (totalOunces / 2.0));     // 12.5
        System.out.println("double / int:     " + (25.0 / cups));           // 12.5
        System.out.println("Remainder:        " + (totalOunces % cups));    // 1
    }
}

SplitTheDrink.main(null);
````

One operand being a `double` was enough to change the entire result. That's the fix for the receipt printer.

> Deliberately converting between types — writing `(double) totalOunces / cups` — is called **casting**, and it's Topic **1.5**. For now, the `2.0` trick is enough.

### Dividing by zero

Back in 1.1 you saw an `ArithmeticException` from dividing an `int` by zero. Now you know exactly which operator caused it.

**Heads up: the next cell is supposed to fail.**

````java
public class NoFriends {
    public static void main(String[] args) {
        int totalOunces = 16;
        int friends     = 0;

        System.out.println("Pouring the drinks...");
        System.out.println("Each person gets: " + (totalOunces / friends));
        System.out.println("This line never runs.");
    }
}

NoFriends.main(null);
````

The first line printed, then the program crashed — the fingerprint of a **run-time error**.

Here's something genuinely surprising, though: `double` division by zero does *not* crash.

````java
public class DoubleDivide {
    public static void main(String[] args) {
        System.out.println("16.0 / 0 = " + (16.0 / 0));    // no crash
        System.out.println("0.0 / 0  = " + (0.0 / 0));     // "not a number"
        System.out.println("Program survived.");
    }
}

DoubleDivide.main(null);
````

`Infinity` and `NaN` ("not a number") are real `double` values Java uses instead of crashing. You won't be asked to compute with them in 1.3 — this is here so the contrast sticks: **`int` division by zero crashes, `double` division by zero doesn't.** You'll revisit precision limits in 1.5.

## Part 6 — Operator precedence

When one expression has several operators, Java doesn't just work left to right. It follows an order.

| Priority | Operators | Direction |
|---|---|---|
| 1 (highest) | `( )` parentheses | innermost first |
| 2 | `*` `/` `%` | left to right |
| 3 (lowest) | `+` `-` | left to right |

Two things students get wrong:

1. **`%` has the same priority as `*` and `/`** — not lower. `10 + 7 % 3` is `10 + 1`, which is `11`.
2. **Same-priority operators go left to right.** `20 / 5 * 2` is `(20 / 5) * 2 = 8`, not `20 / (5 * 2) = 2`.

Parentheses beat everything, and they cost nothing. When in doubt, add them — clear code is worth more than clever code.

````java
public class PrecedencePuzzle {
    public static void main(String[] args) {
        double price    = 5.00;
        int    quantity = 3;

        System.out.println("price * quantity + 2   = " + (price * quantity + 2));    // * first
        System.out.println("price * (quantity + 2) = " + (price * (quantity + 2)));  // parens first

        System.out.println("10 - 2 * 3   = " + (10 - 2 * 3));      // * before -
        System.out.println("20 / 5 * 2   = " + (20 / 5 * 2));      // left to right
        System.out.println("10 + 7 % 3   = " + (10 + 7 % 3));      // % before +
        System.out.println("(10 + 7) % 3 = " + ((10 + 7) % 3));    // parens change everything
    }
}

PrecedencePuzzle.main(null);
````

## Part 7 — The exam skill: tracing output

"What is printed by the following code segment?" Here's the method that works every time.

**Work one statement at a time. For each one: evaluate the expression fully, then decide whether the cursor moves.**

Trace this by hand before you run it:

```java
int a = 7;
int b = 2;
System.out.println("A: " + a / b);
System.out.println("B: " + a % b);
System.out.println("C: " + (a + b) / 3);
System.out.print("D: ");
System.out.print(a * b);
System.out.println(" done");
```

<details>
<summary><b>Worked trace (try it yourself first)</b></summary>

| Statement | Reasoning | Displayed |
|---|---|---|
| `"A: " + a / b` | `/` outranks `+`, so `7 / 2` runs first. Both are `int`, so the result is `3`. Then `"A: " + 3`. | `A: 3` then newline |
| `"B: " + a % b` | `%` outranks `+`. `7 % 2` is `1`. | `B: 1` then newline |
| `"C: " + (a + b) / 3` | Parentheses first: `9`. Then `9 / 3` is `3`. Then concatenate. | `C: 3` then newline |
| `print("D: ")` | No newline — cursor stays put. | `D: ` |
| `print(a * b)` | `14`, still no newline. | `14` on the same line |
| `println(" done")` | Appends, **then** newline. | ` done` |

Final output:

```
A: 3
B: 1
C: 3
D: 14 done
```

The `D:` line is the one people miss — two `print`s and a `println` all land on one line.
</details>

````java
public class TraceMe {
    public static void main(String[] args) {
        int a = 7;
        int b = 2;

        System.out.println("A: " + a / b);
        System.out.println("B: " + a % b);
        System.out.println("C: " + (a + b) / 3);
        System.out.print("D: ");
        System.out.print(a * b);
        System.out.println(" done");
    }
}

TraceMe.main(null);
````

**(B) 18**. `/` and `*` share the top priority and run left to right: `5 / 2` is `2` (int division), then `2 * 3` is `6`, then `12 + 6` is `18`. Answer (A) comes from doing `5 / 2` as `2.5`; answer (C) comes from evaluating `2 * 3` before the division.
</details>

**7.** A program computes `int average = totalOunces / cups;` where `totalOunces` is 25 and `cups` is 2. The program compiles, runs, and prints `12`. This is an example of a:

&nbsp;&nbsp;(A) Compile-time error &nbsp;&nbsp; (B) Run-time error &nbsp;&nbsp; (C) Logic error &nbsp;&nbsp; (D) Correct program

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| `System.out.print` | Displays output and leaves the cursor on the same line |
| `System.out.println` | Displays output, then moves the cursor to a new line |
| String literal | Fixed text written inside double quotation marks |
| Escape sequence | A backslash pair like `\"`, `\\`, or `\n` that represents a special character |
| Concatenation | Using `+` to join a String with another value, converting it to text |
| Integer division | `int / int` produces an `int`; the fractional part is discarded, not rounded |
| Remainder operator `%` | Gives what's left over after integer division |
| Operator precedence | `( )` first, then `*` `/` `%`, then `+` `-`; ties break left to right |

### The five things that will show up on the exam

1. `print` leaves the cursor; `println` moves it.
2. `+` chains evaluate **left to right** — once a String appears, everything after it concatenates.
3. `int / int` **truncates**. `7 / 2` is `3`.
4. One `double` operand makes the whole result a `double`.
5. `%` has the **same** priority as `*` and `/`, not lower.
