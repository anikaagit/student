## Part 1 — What is a variable?

> **Definition:** A **variable** is a named location in memory that stores a value.

Picture the back counter at the cafe. There's a row of labeled containers: one marked *pearls*, one marked *brown sugar syrup*, one marked *oat milk*. Three facts about those containers are true of variables too:

1. Each has a **name** (the label) so you can find it again.
2. Each holds **one thing at a time**. Pour something new in and the old contents are gone.
3. Each has a **type** — the pearl container holds pearls, not syrup. You can't mix them up.

That third one is the big idea of this lesson. Java is **statically typed**: every variable must be given a type, and it holds that type forever.

Run the cell below. Watch the value change when we assign a new one.

````java
public class InventoryBoard {
    public static void main(String[] args) {
        int pearlServings = 40;
        System.out.println("Pearl servings left: " + pearlServings);

        pearlServings = 39;          // a customer just ordered -- the old 40 is gone
        System.out.println("Pearl servings left: " + pearlServings);

        pearlServings = 38;
        System.out.println("Pearl servings left: " + pearlServings);
    }
}

InventoryBoard.main(null);
````

The container `pearlServings` never held 40 *and* 39. Each assignment **replaced** what was there.

## Part 2 — Declaration, initialization, assignment

These three words get used loosely in conversation, but on the AP exam they mean specific things. Learn the difference now.

| Term | What it does | Example |
|---|---|---|
| **Declaration** | Creates the variable: reserves memory and gives it a name and a type | `int cupsSold;` |
| **Assignment** | Puts a value into an existing variable, using the `=` operator | `cupsSold = 12;` |
| **Initialization** | The *first* assignment — often combined with the declaration | `int cupsSold = 12;` |

One thing that surprises people: `=` does **not** mean "equals" in the math sense. It means **"take the value on the right and store it in the variable on the left."** It's an arrow pointing left, not a statement of fact.

That's why this line, which is nonsense in algebra, is completely normal in Java:

```java
cupsSold = cupsSold + 1;
```

Read it right-to-left: *take the current value of `cupsSold`, add 1, and store the result back into `cupsSold`.*

````java
public class ThreeSteps {
    public static void main(String[] args) {
        int cupsSold;                      // DECLARATION -- container exists, but it's empty
        cupsSold = 12;                     // ASSIGNMENT  -- now it holds 12
        System.out.println("Cups sold: " + cupsSold);

        double dailyTotal = 138.50;        // DECLARATION + INITIALIZATION in one line
        System.out.println("Daily total: $" + dailyTotal);

        cupsSold = cupsSold + 1;           // read right-to-left: 12 + 1 = 13, stored back
        System.out.println("Cups sold now: " + cupsSold);
    }
}

ThreeSteps.main(null);
````

### What if you skip the assignment?

Declaring a variable does not put anything in it. Java refuses to let you read an empty container.

**Heads up: the next cell is supposed to fail.** Run it and read the message.

````java
public class TooSoon {
    public static void main(String[] args) {
        int strawCount;                                    // declared, but never assigned
        System.out.println("Straws left: " + strawCount);  // trying to read an empty container
    }
}

TooSoon.main(null);
````

The compiler says the variable *might not have been initialized*. Note the error type from 1.1: this is a **compile-time error**, so nothing ran at all.

## Part 3 — The three primitive types

> **Definition:** A **data type** is a set of possible values plus the operations you can perform on them.

Java has eight primitive types, but **AP CSA only uses three**. These are the ones you'll be tested on:

| Type | Holds | Example values | Cafe use |
|---|---|---|---|
| `int` | Whole numbers, positive or negative, **no decimal point** | `40`, `-3`, `0` | Cups sold, sugar level %, pearl scoops |
| `double` | Real numbers **with** a decimal point | `5.75`, `-0.5`, `3.0` | Prices, tax, ounces of milk |
| `boolean` | Exactly two values: `true` or `false` | `true`, `false` | Is the shop open? Dairy free? |

Notice all three are written in **lowercase**. That's a reliable clue that a type is primitive — you'll see why in Part 5.

A `boolean` may look like the least useful of the three, but it's what makes selection work. Every `if` condition you write in Unit 2 evaluates to a `boolean`.

````java
public class MenuData {
    public static void main(String[] args) {
        int     sugarLevel  = 50;        // a whole percent -- 0, 25, 50, 75, 100
        double  price       = 5.75;      // dollars and cents
        boolean isDairyFree = true;      // a yes/no fact

        System.out.println("Sugar level:  " + sugarLevel + "%");
        System.out.println("Price:        $" + price);
        System.out.println("Dairy free?   " + isDairyFree);
    }
}

MenuData.main(null);
````

### The promise from Lesson 1: why a price is never an `int`

Last notebook ended by claiming you should never store a price as an `int`. Here's the proof.

**Heads up: the next cell is supposed to fail.**

````java
public class PriceProblem {
    public static void main(String[] args) {
        int price = 5.75;                       // an int container cannot hold a decimal
        System.out.println("Price: $" + price);
    }
}

PriceProblem.main(null);
````

Read that error closely: **possible lossy conversion from double to int**. Java is telling you that squeezing `5.75` into an `int` would silently throw away the `.75` — and it refuses to lose your customer's money without permission.

This is Java protecting you. A weaker language might have stored `5` and let you find out at the register.

````java
public class PriceFixed {
    public static void main(String[] args) {
        double price = 5.75;
        System.out.println("Price: $" + price);

        // A preview of a famous double quirk you'll meet again in 1.5:
        System.out.println("0.1 + 0.2 = " + (0.1 + 0.2));
    }
}

PriceFixed.main(null);
````

That last line is not a bug in Java. A `double` stores an **approximation** of a real number in binary, and some decimals can't be represented exactly — the same way `1/3` can't be written exactly in decimal. It's usually invisible, but it's the reason serious money software often stores prices as a whole number of **cents** in an `int` instead. For this course, `double` is the right answer for a price.

### booleans in action

````java
public class ShopStatus {
    public static void main(String[] args) {
        boolean isOpen  = true;
        boolean hasTaro = false;

        System.out.println("Shop open?     " + isOpen);
        System.out.println("Taro in stock? " + hasTaro);

        // && means "and" -- true only if BOTH are true. More on this in Unit 2.
        System.out.println("Can we serve taro right now? " + (isOpen && hasTaro));
    }
}

ShopStatus.main(null);
````

## Part 4 — Literals

> **Definition:** A **literal** is a value written directly in your source code.

In `int sugarLevel = 50;`, the `50` is an int literal. Literals carry a type all on their own, and Java can tell them apart by how they're written:

| Written as | Literal type |
|---|---|
| `7` | `int` |
| `7.0` or `7.5` | `double` — the decimal point is what makes it a double |
| `true` / `false` | `boolean` |
| `"7"` | `String` — quotation marks mean **text**, not a number |

The difference between `7` and `"7"` matters enormously. One is a number you can do math with; the other is a character sitting in a piece of text. Run the cell and watch the output of `7` and `7.0` differ.

````java
public class LiteralTour {
    public static void main(String[] args) {
        System.out.println(7);       // int literal
        System.out.println(7.0);     // double literal -- prints with the .0 attached
        System.out.println(true);    // boolean literal
        System.out.println("7");     // String literal -- this is TEXT
        System.out.println("Byte-Sized Boba");
    }
}

LiteralTour.main(null);
````

## Part 5 — `String`: your first reference type

Java splits every data type into two families.

| | **Primitive types** | **Reference types** |
|---|---|---|
| Written | lowercase: `int`, `double`, `boolean` | Capitalized: `String`, `Scanner`, `Math` |
| The variable holds | the **actual value** | a **reference** (memory address) pointing to an object stored elsewhere |
| Cafe analogy | the container holds the pearls | the container holds a **claim ticket** for a bag in the back room |

Here's the picture in memory:

```
  PRIMITIVE                          REFERENCE
  int cupsSold = 12;                 String name = "Riley";

  cupsSold                           name
 ┌──────────┐                       ┌──────────┐        ┌───────────┐
 │    12    │  <- the value          │  0x4a2f  │ ─────► │  "Riley"  │
 └──────────┘     lives right here   └──────────┘        └───────────┘
                                      an address          the actual
                                      to somewhere         object
```

`String` is a **class**, not a primitive — which is why it gets a capital `S`. That capital letter is your everyday tip-off.

> You'll dig into `String` methods in **1.3** and into objects generally in **1.6**. For now: know that it's a reference type and that it's capitalized.

````java
public class DrinkLabel {
    public static void main(String[] args) {
        String  customerName = "Riley";           // reference type -- capital S
        String  drinkName    = "Taro Milk Tea";
        int     quantity     = 2;                 // primitive
        double  unitPrice    = 5.75;              // primitive
        boolean extraPearls  = true;              // primitive

        System.out.println("--- ORDER LABEL ---");
        System.out.println("Name:   " + customerName);
        System.out.println("Drink:  " + quantity + "x " + drinkName);
        System.out.println("Each:   $" + unitPrice);
        System.out.println("Pearls: " + extraPearls);
    }
}

DrinkLabel.main(null);
````

## Part 6 — Naming your variables

The name of a variable is called an **identifier**. Java has hard *rules* (break them and it won't compile) and soft *conventions* (break them and it compiles, but your teacher and every Java programmer alive will wince).

### The rules — required

- Must begin with a **letter**, an underscore `_`, or a dollar sign `$`
- **Cannot** begin with a digit
- May contain letters, digits, `_`, and `$` after the first character
- **No spaces** and no other symbols
- Cannot be a **reserved word** (`class`, `int`, `public`, `static`, `if`, `for`, `true`, ...)
- Case sensitive: `price`, `Price`, and `PRICE` are three different variables

### The conventions — expected

- Use **camelCase** for variables: start lowercase, capitalize each new word — `sugarLevel`, `customerName`, `isDairyFree`
- Use **ALL_CAPS_WITH_UNDERSCORES** for constants — `TAX_RATE`, `MAX_PEARLS`
- Use **PascalCase** for class names — `DrinkLabel`, `MenuData`
- Make the name say what it holds. `x` and `temp1` tell a reader nothing; `pearlServings` tells them everything.
- `boolean` names read nicely as questions: `isOpen`, `hasTaro`, `wantsIce`

**Heads up: the next cell is supposed to fail — it breaks three rules at once.**

````java
public class BadNames {
    public static void main(String[] args) {
        int 2ndDrinkPrice = 6;      // cannot start with a digit
        double class = 4.50;        // "class" is a reserved word
        int drink price = 3;        // no spaces allowed in an identifier
    }
}

BadNames.main(null);
````

````java
public class GoodNames {
    public static void main(String[] args) {
        int     secondDrinkPrice = 6;         // camelCase, starts with a letter
        double  toppingCost      = 0.75;      // says exactly what it holds
        boolean isLargeSize      = true;      // reads like a question
        String  customerName     = "Riley";

        System.out.println(customerName + " ordered a large? " + isLargeSize);
        System.out.println("Second drink: $" + secondDrinkPrice + ", topping: $" + toppingCost);
    }
}

GoodNames.main(null);
````

## Part 7 — Constants with `final`

Some values should never change while the program runs. The sales tax rate. The maximum pearl scoops per drink. The name of the shop.

The `final` keyword locks a variable after it's initialized. Try to change it later and the **compiler** stops you — you find the mistake before the program ever runs, instead of after a customer gets overcharged.

Constants are conventionally written in `ALL_CAPS_WITH_UNDERSCORES` and declared at the top of the class as `public static final`.

````java
public class CafeConstants {

    public static final double TAX_RATE   = 0.0775;   // constants live at the top
    public static final int    MAX_PEARLS = 4;
    public static final String SHOP_NAME  = "Byte-Sized Boba";

    public static void main(String[] args) {
        double subtotal = 11.50;
        double total    = subtotal + subtotal * TAX_RATE;

        System.out.println("Welcome to " + SHOP_NAME);
        System.out.println("Subtotal: $" + subtotal);
        System.out.println("Total:    $" + total);
        System.out.println("Max pearl scoops per drink: " + MAX_PEARLS);
    }
}

CafeConstants.main(null);
````

Notice the total printed as `$12.39125`. Storing the value correctly and *displaying* it nicely are two different jobs — `System.out.printf("$%.2f%n", total)` from Lesson 1.1 handles the display side. Right now we only care that the number in memory is right.

Why bother with `final`? Two reasons worth remembering:

1. **Safety.** A typo that would silently change your tax rate becomes a compile error instead.
2. **One place to edit.** If the tax rate changes, you fix one line — not the fourteen places you typed `0.0775`.

**Heads up: the next cell is supposed to fail.**

````java
public class ChangeTheTax {

    public static final double TAX_RATE = 0.0775;

    public static void main(String[] args) {
        System.out.println("Tax rate: " + TAX_RATE);
        TAX_RATE = 0.05;                   // trying to overwrite a final variable
        System.out.println("New tax rate: " + TAX_RATE);
    }
}

ChangeTheTax.main(null);
````

## Part 8 — Choosing the right type

Most type mistakes come from not asking the right question. Use this:

```
Is it text (a name, a flavor, a message)?          --> String
Is it a yes/no fact?                               --> boolean
Is it a number?
      Could it ever need a decimal point?          --> double
      Is it always a whole count?                  --> int
```

Worked examples from the cafe:

| Data | Type | Why |
|---|---|---|
| Number of cups in the stack | `int` | You can't have 3.5 cups |
| Price of a drink | `double` | $5.75 needs the decimal |
| Sugar level percentage | `int` | The menu only offers 0, 25, 50, 75, 100 |
| Ounces of milk poured | `double` | 6.5 oz is a real measurement |
| Customer's name | `String` | Text |
| Whether the customer wants ice | `boolean` | Yes or no, nothing in between |
| Loyalty points earned | `int` | Points are whole |
| Store's sales tax rate | `double` + `final` | A decimal that must never change |

One trap: **a number written as text is not a number.** A phone number is text (`String`) even though it's made of digits — you're never going to do arithmetic on it, and it might start with a `0` that an `int` would erase.

| Declaration | Creating a variable by giving it a type and a name |
| Assignment | Storing a value into a variable using `=` |
| Initialization | The first assignment given to a variable |
| `int` | Primitive type for whole numbers |
| `double` | Primitive type for numbers with a decimal point |
| `boolean` | Primitive type holding only `true` or `false` |
| Primitive type | Stores the actual value; written lowercase |
| Reference type | Stores a memory address pointing to an object; written capitalized |
| `String` | A reference type that holds text |
| Literal | A value written directly in source code |
| Identifier | The name of a variable, method, or class |
| camelCase | Naming convention for variables: `sugarLevel` |
| `final` | Keyword making a variable a constant that cannot be reassigned |
