## Part 1 — The five compound operators

Each one performs an operation on a variable and stores the result **back into that same variable**.

| Operator | Example | Long form | Meaning |
|---|---|---|---|
| `+=` | `total += 5;` | `total = total + 5;` | Add and store |
| `-=` | `stock -= 1;` | `stock = stock - 1;` | Subtract and store |
| `*=` | `total *= 2;` | `total = total * 2;` | Multiply and store |
| `/=` | `total /= 4;` | `total = total / 4;` | Divide and store |
| `%=` | `pearls %= 5;` | `pearls = pearls % 5;` | Take the remainder and store |

Everything you learned in 1.4 still applies: the right side is evaluated first, then the result is stored, overwriting the old value.

**The critical thing they all have in common:** they **store**. Writing `total + 5;` on its own computes a value and throws it away. `total += 5;` actually changes `total`.

````java
public class CompoundTour {
    public static void main(String[] args) {
        double total = 10.00;
        System.out.println("Start:            $" + total);

        total += 5.75;
        System.out.println("total += 5.75 ->  $" + total);

        total -= 2.00;
        System.out.println("total -= 2.00 ->  $" + total);

        total *= 2;
        System.out.println("total *= 2    ->  $" + total);

        total /= 4;
        System.out.println("total /= 4    ->  $" + total);
    }
}

CompoundTour.main(null);
````

````java
public class ModCompound {
    public static void main(String[] args) {
        int pearls        = 17;    // scoops left in the container
        int scoopsPerDrink = 5;

        System.out.println("Scoops available: " + pearls);

        pearls %= scoopsPerDrink;   // same as: pearls = pearls % scoopsPerDrink

        System.out.println("Scoops left after making full drinks: " + pearls);
    }
}

ModCompound.main(null);
````

## Part 2 — The implicit parentheses rule

Here's the first trap, and it's a good one.

> **`x op= expression` means `x = x op (expression)`.** The entire right side gets **invisible parentheses** around it.

That's not the same as just pasting the expression into the long form. Compare:

```java
a *= 2 + 3;        // means  a = a * (2 + 3)   -> a * 5
b = b * 2 + 3;     // means  b = (b * 2) + 3   -> different!
```

The compound version wraps the whole right side. The long-form version obeys normal precedence from 1.3, where `*` beats `+`.

Starting from `10`, one gives `50` and the other gives `23`.

````java
public class ImplicitParens {
    public static void main(String[] args) {
        int a = 10;
        a *= 2 + 3;                  // a = a * (2 + 3)  =  10 * 5
        System.out.println("a *= 2 + 3      ->  " + a);

        int b = 10;
        b = b * 2 + 3;               // b = (b * 2) + 3  =  20 + 3
        System.out.println("b = b * 2 + 3   ->  " + b);

        int c = 20;
        c /= 2 + 3;                  // c = c / (2 + 3)  =  20 / 5
        System.out.println("c /= 2 + 3      ->  " + c);
    }
}

ImplicitParens.main(null);
````

So when you translate a compound statement to its long form on an exam, **add the parentheses yourself**:

```java
total *= quantity + 1;
// long form:  total = total * (quantity + 1);
// NOT:        total = total * quantity + 1;
```

## Part 3 — `/=` and `%=` on integers

The second trap. Compound operators don't change the type rules — they inherit them.

> **If the variable is an `int`, `/=` performs integer division and truncates.**

This is the same rule from 1.3 and 1.5, just wearing a shorter costume. `cups /= 2` on an `int` throws away the fraction exactly like `cups / 2` would.

````java
public class IntegerCompound {
    public static void main(String[] args) {
        int cupsInt = 7;
        cupsInt /= 2;
        System.out.println("int    7 /= 2  ->  " + cupsInt);

        double cupsDouble = 7;
        cupsDouble /= 2;
        System.out.println("double 7 /= 2  ->  " + cupsDouble);
    }
}

IntegerCompound.main(null);
````

Same operator, same numbers, different answers — decided entirely by the **declared type of the variable on the left**.

### Truncation compounds

Do it repeatedly and the losses stack up. Watch a stock count of 100 get divided by 3 three times:

````java
public class RepeatedTruncation {
    public static void main(String[] args) {
        int stock = 100;
        System.out.println("Start:      " + stock);

        stock /= 3;
        System.out.println("After /= 3: " + stock);     // 33, not 33.33...

        stock /= 3;
        System.out.println("After /= 3: " + stock);     // 11, from 33

        stock /= 3;
        System.out.println("After /= 3: " + stock);     // 3, from 11

        System.out.println();
        System.out.println("True value of 100 / 27 is about 3.70");
    }
}

RepeatedTruncation.main(null);
````

Mathematically, dividing by 3 three times is dividing by 27, and `100 / 27` is about `3.70`. Java gave `3`. Each step quietly discarded a fraction, and the errors accumulated.

**The fix, as always from 1.5: use a `double` variable if you need the fraction.** No cast can rescue precision after the fact.

## Part 4 — `++` and `--`

Adding or subtracting exactly 1 is so common that Java has a dedicated operator for it.

| Operator | Name | Effect |
|---|---|---|
| `++` | Increment | Adds 1 to the variable |
| `--` | Decrement | Subtracts 1 from the variable |

So these three statements all do the same thing:

```java
cupsSold = cupsSold + 1;
cupsSold += 1;
cupsSold++;
```

> **On the AP exam, `++` and `--` appear as standalone statements** — a complete statement on their own line, or in the update section of a `for` loop header. The result isn't used as part of a larger expression, so you don't need to worry about anything subtle.

````java
public class CupCounter {
    public static void main(String[] args) {
        int cupsSold  = 0;
        int cupsStock = 50;

        System.out.println("Opening -- sold: " + cupsSold + ", stock: " + cupsStock);

        cupsSold++;    cupsStock--;
        cupsSold++;    cupsStock--;
        cupsSold++;    cupsStock--;

        System.out.println("After 3 sales -- sold: " + cupsSold + ", stock: " + cupsStock);

        cupsSold--;    cupsStock++;      // a refund
        System.out.println("After refund  -- sold: " + cupsSold + ", stock: " + cupsStock);
    }
}

CupCounter.main(null);
````

You'll see `i++` constantly starting in Unit 2:

```java
for (int i = 0; i < 5; i++) {     // i++ advances the counter each pass
    System.out.println("Sealing cup " + i);
}
```

That's the form that shows up in nearly every array and loop question on the exam.

---

### Beyond the exam: prefix vs. postfix

You may run into `++x` (prefix) written before the variable instead of after. **As a standalone statement, `x++` and `++x` are identical** — both just add 1.

They only differ when the result is used *inside a larger expression*: `x++` hands back the **old** value and then increments; `++x` increments **first** and hands back the new value.

**The AP exam does not test this**, and the College Board deliberately keeps increment operators out of larger expressions. It's here so the code doesn't confuse you if you meet it in the wild — and because writing it is a good way to make code unreadable.

````java
public class PrefixPostfix {
    public static void main(String[] args) {
        int a = 5;
        int resultPost = a++;      // POSTFIX: hand back 5, THEN make a into 6
        System.out.println("int resultPost = a++;  ->  result = " + resultPost + ", a = " + a);

        int b = 5;
        int resultPre = ++b;       // PREFIX: make b into 6 FIRST, then hand back 6
        System.out.println("int resultPre  = ++b;  ->  result = " + resultPre + ", b = " + b);

        System.out.println();
        System.out.println("Both variables ended at 6. Only the returned value differed.");
    }
}

PrefixPostfix.main(null);
````

## Part 5 — A full shift, in shorthand

Everything together. Compare this to the version you'd have written in Lesson 1.4 — same logic, noticeably less noise.

````java
public class ShiftTotals {

    public static final double TAX_RATE = 0.0775;

    public static void main(String[] args) {
        double revenue    = 0.0;
        int    drinksSold = 0;
        int    cupsStock  = 50;

        // Milk tea
        revenue += 5.75;   drinksSold++;   cupsStock--;
        // Matcha latte
        revenue += 6.50;   drinksSold++;   cupsStock--;
        // Thai tea
        revenue += 4.25;   drinksSold++;   cupsStock--;

        System.out.println("Drinks sold:   " + drinksSold);
        System.out.println("Cups in stock: " + cupsStock);
        System.out.println("Subtotal:      $" + revenue);

        revenue += revenue * TAX_RATE;     // revenue = revenue + (revenue * TAX_RATE)

        System.out.println("With tax:      $" + revenue);
    }
}

ShiftTotals.main(null);
````

That last line is worth a second look. `revenue += revenue * TAX_RATE` means `revenue = revenue + (revenue * TAX_RATE)` — the implicit parentheses from Part 2 wrap the whole multiplication, which is exactly what you want here.

## Part 6 — Tracing compound assignments

Same technique as 1.4: one row per statement, evaluate the right side fully, then store.

Trace this before running it:

```java
int cups = 24;
cups /= 4;
cups += 3;
cups *= 2;
cups %= 7;
cups--;
```

<details>
<summary><b>Worked trace (try it first)</b></summary>

| Statement | Long form | Arithmetic | `cups` |
|---|---|---|---|
| `int cups = 24;` | | | **24** |
| `cups /= 4;` | `cups = cups / 4` | `24 / 4` = `6` | **6** |
| `cups += 3;` | `cups = cups + 3` | `6 + 3` = `9` | **9** |
| `cups *= 2;` | `cups = cups * 2` | `9 * 2` = `18` | **18** |
| `cups %= 7;` | `cups = cups % 7` | `18 % 7` = `4` | **4** |
| `cups--;` | `cups = cups - 1` | `4 - 1` = `3` | **3** |

Final: `cups` is `3`.

The `%=` step is the one to watch — 7 goes into 18 twice with **4** left over.
</details>

````java
public class TraceCompound {
    public static void main(String[] args) {
        int cups = 24;

        cups /= 4;
        cups += 3;
        cups *= 2;
        cups %= 7;
        cups--;

        System.out.println("cups = " + cups);
    }
}

TraceCompound.main(null);
````

**Bugs 1 and 2 — the operators are backwards.** `=+` and `=-` are not operators at all. Java reads `drinksSold =+ 1;` as `drinksSold = +1;` — a plain assignment of positive one. Likewise `cupsInStock =- 1;` is `cupsInStock = -1;`.

# Self-Check: AP-style questions

**1.** What is the value of `x` after this code segment?

```java
int x = 20;
x -= 5;
x /= 3;
```

&nbsp;&nbsp;(A) `5` &nbsp;&nbsp; (B) `15` &nbsp;&nbsp; (C) `4` &nbsp;&nbsp; (D) `6.67`

<details><summary>Answer</summary>

**(A) 5**. `x -= 5` gives `15`, then `x /= 3` gives `5`. Answer (D) forgets that `x` is an `int` and that the value would be stored as an `int` regardless.
</details>

**2.** Which statement is equivalent to `total *= price + 2;`?

&nbsp;&nbsp;(A) `total = total * price + 2;`
&nbsp;&nbsp;(B) `total = total * (price + 2);`
&nbsp;&nbsp;(C) `total = (total * price) + 2;`
&nbsp;&nbsp;(D) `total = price + 2;`

<details><summary>Answer</summary>

**(B)**. A compound operator places implicit parentheses around the **entire** right side. (A) and (C) are the same thing as each other and both apply normal precedence, which is not what the compound form does.
</details>

**3.** What is the value of `cups` after this code segment?

```java
int cups = 9;
cups /= 2;
cups *= 2;
```

&nbsp;&nbsp;(A) `9` &nbsp;&nbsp; (B) `8` &nbsp;&nbsp; (C) `10` &nbsp;&nbsp; (D) `4`

<details><summary>Answer</summary>

**(B) 8**. `9 / 2` is integer division giving `4` — the `.5` is gone permanently. Then `4 * 2` gives `8`. Dividing and re-multiplying does **not** return you to the original value once truncation has happened.
</details>

**4.** A programmer writes `count =- 1;` intending to subtract 1 from `count`. What actually happens?

&nbsp;&nbsp;(A) A compile-time error
&nbsp;&nbsp;(B) `count` is reduced by 1 as intended
&nbsp;&nbsp;(C) `count` is assigned the value `-1`
&nbsp;&nbsp;(D) A run-time error

<details><summary>Answer</summary>

**(C)**. Java reads it as `count = -1;` — an assignment of negative one. It compiles cleanly and produces no warning, making it a **logic error** that can be very hard to spot.
</details>

**6.** Which of the following does **not** change the value of `total`?

&nbsp;&nbsp;(A) `total += 0;`
&nbsp;&nbsp;(B) `total *= 1;`
&nbsp;&nbsp;(C) `total + 5;`
&nbsp;&nbsp;(D) `total -= 0;`

<details><summary>Answer</summary>

**(C)**. It computes a value and discards it, because there's no assignment — nothing is ever stored back into `total`. (A), (B), and (D) all perform an assignment; they just happen to store the same value that was already there.
</details>

**7.** Given `double d = 9;` followed by `d /= 2;`, and `int i = 9;` followed by `i /= 2;`, what are the values of `d` and `i`?

&nbsp;&nbsp;(A) `d = 4.5`, `i = 4`
&nbsp;&nbsp;(B) `d = 4.5`, `i = 4.5`
&nbsp;&nbsp;(C) `d = 4.0`, `i = 4`
&nbsp;&nbsp;(D) `d = 4.5`, `i = 5`

<details><summary>Answer</summary>

**(A)**. The declared type of the variable on the left decides everything. `d` is a `double`, so the division keeps the fraction. `i` is an `int`, so `/=` truncates. An `int` variable can never hold `4.5`, which rules out (B).
</details>

**8.** In a `for` loop header written as `for (int i = 0; i < n; i++)`, why does it not matter whether the programmer writes `i++` or `++i`?

&nbsp;&nbsp;(A) Java treats them as the same operator
&nbsp;&nbsp;(B) The result is not used in a larger expression, so both simply add 1
&nbsp;&nbsp;(C) `++i` is not valid Java
&nbsp;&nbsp;(D) Loop headers ignore increment operators

<details><summary>Answer</summary>

**(B)**. Prefix and postfix differ only in the value they hand back to a surrounding expression. In a loop header the returned value is discarded, so both forms have the identical effect of incrementing `i`.
</details>

# Closing time

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Compound assignment operator | `+=`, `-=`, `*=`, `/=`, `%=` — operate on a variable and store the result back |
| Implicit parentheses | A compound operator wraps its entire right side: `x *= a + b` means `x = x * (a + b)` |
| Increment operator `++` | Adds 1 to a variable |
| Decrement operator `--` | Subtracts 1 from a variable |
| Standalone statement | An increment or decrement on its own, not embedded in a larger expression |

### The five things that will show up on the exam

1. `x += y` is exactly `x = x + y`.
2. The right side gets **implicit parentheses**: `x *= 2 + 3` means `x = x * 5`.
3. `/=` on an `int` **truncates**, same as `/`.
4. `x++` and `x--` add or subtract 1 as standalone statements.
5. `total + 5;` computes nothing useful — only an assignment **stores**.

### The typo to never make

```java
count =- 1;    // assigns -1.  Compiles fine. Wrecks your program.
count -= 1;    // subtracts 1. What you meant.
```
