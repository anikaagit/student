## Part 1 — Libraries

> **Definition:** A **library** is a collection of pre-written classes that you can use in your own programs.

Java's **standard library** comes bundled with Java itself. You don't install it, download it, or pay for it — it's already there, which is why `System.out.println` worked on day one without any setup.

Three reasons this matters more than it might sound:

| Reason | The cafe version |
|---|---|
| **Don't reinvent** | You don't build a milk steamer from scratch to open a cafe |
| **Already tested** | The supplier's steamer has been used by ten thousand shops; yours has been used by you |
| **Everyone knows it** | Any barista you hire already knows how a standard steamer works |

That last one is easy to underestimate. When you write `Math.sqrt(x)`, every Java programmer on earth instantly knows what that does. If you wrote your own `mySquareRootThing(x)`, nobody would — including you, six months later.

---

## Part 2 — What an API actually is

**API** stands for **Application Program Interface**.

> **Definition:** An **API** is the documented description of what a class provides and how to use it — its available methods, what they need, and what they give back.

Here's the part that trips people up: **an API tells you what a class does, not how it does it.**

Think about the cafe menu. It tells a customer:

- what they can order (Taro Milk Tea)
- what they have to supply (a size choice, $5.75)
- what they'll get back (a sealed cup of taro milk tea)

It does **not** tell them the steeping time, the brand of tapioca, or how the sealing machine works. The customer doesn't need any of that — and the cafe can switch tapioca suppliers without reprinting a single menu.

That separation is called **abstraction**, and it's one of the most important ideas in the whole course.

```
   WHAT YOU SEE (the API)              WHAT'S HIDDEN (the implementation)
 ┌──────────────────────────┐        ┌────────────────────────────────┐
 │  Math.sqrt(double x)     │        │  hundreds of lines of numerical│
 │  Returns the square root │───────►│  approximation code you never  │
 │  of x                    │        │  have to read or understand    │
 └──────────────────────────┘        └────────────────────────────────┘
```

You have used `Math.sqrt` with total confidence and zero idea how square roots are computed in binary. That's not laziness — **that's the API doing its job.**

````java
public class Abstraction {
    public static void main(String[] args) {
        // You know exactly WHAT this does.
        // You have no idea HOW, and you don't need to.
        System.out.println("Square root of 144: " + Math.sqrt(144));
        System.out.println("Square root of 2:   " + Math.sqrt(2));

        System.out.println();
        System.out.println("Two calls. Zero knowledge of the algorithm required.");
    }
}

Abstraction.main(null);
````

## Part 3 — Attributes and behaviors

This is the actual learning objective, so slow down here.

Every class describes two kinds of things:

| | **Attributes** | **Behaviors** |
|---|---|---|
| Also called | Data, state, fields, variables | Methods, operations |
| Question it answers | *What does it **know**?* | *What can it **do**?* |
| In code | Variables belonging to the class | Methods belonging to the class |

For a drink at the cafe:

```
        BobaDrink
 ┌───────────────────────────┐
 │ ATTRIBUTES (what it knows)│
 │   flavor      "Taro"      │
 │   sizeOunces  16          │
 │   sugarLevel  50          │
 │   hasPearls   true        │
 │   price       5.75        │
 ├───────────────────────────┤
 │ BEHAVIORS (what it does)  │
 │   addTopping()            │
 │   setSugarLevel()         │
 │   getPrice()              │
 │   seal()                  │
 └───────────────────────────┘
```

A quick test that almost always works: **attributes are nouns, behaviors are verbs.** `price` is a thing the drink knows. `seal()` is a thing the drink does.

You can spot both in classes you've already used:

````java
public class AttributesAndBehaviors {
    public static void main(String[] args) {
        // ATTRIBUTES -- values the class KNOWS. No parentheses.
        System.out.println("Integer.MAX_VALUE  (attribute): " + Integer.MAX_VALUE);
        System.out.println("Integer.MIN_VALUE  (attribute): " + Integer.MIN_VALUE);

        System.out.println();

        // BEHAVIORS -- things the class DOES. Parentheses, usually with input.
        System.out.println("Math.abs(-7)       (behavior):  " + Math.abs(-7));
        System.out.println("Math.pow(2, 5)     (behavior):  " + Math.pow(2, 5));
        System.out.println("Math.sqrt(144)     (behavior):  " + Math.sqrt(144));
    }
}

AttributesAndBehaviors.main(null);
````

Notice the syntax difference, which is a reliable giveaway:

- `Integer.MAX_VALUE` — **no parentheses**. You're reading a stored value.
- `Math.sqrt(144)` — **parentheses**, with something inside. You're asking it to *do* something.

> You'll write your own classes with your own attributes and behaviors in **Unit 3**. Right now you only need to identify them in classes somebody else wrote.

## Part 4 — Packages and `import`

Java's standard library is enormous, so it's organized into **packages** — like aisles in a supply warehouse.

| Package | Contains | Import needed? |
|---|---|---|
| `java.lang` | `String`, `Math`, `System`, `Integer`, `Double`, `Object` | **No** — automatic |
| `java.util` | `Scanner`, `ArrayList` | **Yes** |
| `java.io` | `File` | **Yes** |

`java.lang` is imported into every Java program automatically, because its contents are so fundamental. That's the answer to the question at the top of this notebook: **`System` worked on day one because `System` lives in `java.lang`.**

Everything else you have to ask for:

```java
import java.util.Scanner;        // one specific class
import java.util.*;              // every class in the package
```

The `import` goes **above** the class declaration, at the very top of the file. Importing a class you never use is harmless; using a class you never imported is a compile-time error.

````java
import java.util.Scanner;

public class ProperImport {
    public static void main(String[] args) {
        // Scanner is from java.util, so the import above is required
        Scanner scan = new Scanner("Taro 3 5.75");

        System.out.println("Flavor:   " + scan.next());
        System.out.println("Quantity: " + scan.nextInt());
        System.out.println("Price:    $" + scan.nextDouble());

        scan.close();

        // Math and Integer are from java.lang -- no import, they just work
        System.out.println("Also available with no import: " + Math.sqrt(64));
    }
}

ProperImport.main(null);
````

### What forgetting the import looks like

If you write this in a real `.java` file:

```java
public class NoImport {
    public static void main(String[] args) {
        Scanner scan = new Scanner("Taro 3");     // no import above!
    }
}
```

the compiler stops you before anything runs:

```
NoImport.java:3: error: cannot find symbol
        Scanner scan = new Scanner("Taro 3");
        ^
  symbol:   class Scanner
  location: class NoImport
```

**"Cannot find symbol" on a class name almost always means a missing import.** It's one of the most common errors beginners hit, and once you recognize the message it takes five seconds to fix.

> **An honest caveat about this notebook.** Jupyter's Java kernel automatically pre-imports several common packages, including `java.util`, as a convenience. That means a cell using `Scanner` without an `import` may run just fine *here* while failing in a real `.java` file.
>
> This is why the example above is shown as text rather than as a runnable cell — it would not reliably fail for you. **Always write the `import` anyway.** Your notebook is being generous; the AP exam and every real compiler will not be.

## Part 5 — How to read an API entry

Documentation entries look intimidating until you know they always have the same four parts.

```
   static double pow(double base, double exponent)
   ^^^^^^ ^^^^^^ ^^^  ^^^^^^^^^^^^^^^^^^^^^^^^^^^^
     |      |     |              |
     |      |     |              └─ PARAMETERS: what you must supply
     |      |     └─ NAME: what you call it
     |      └─ RETURN TYPE: what type comes back
     └─ static: call it on the class itself (Math.pow), not on an object
```

Read it as a sentence: *"`pow` needs two doubles, and gives you back a double."*

Then the description tells you the meaning: *"Returns the value of the first parameter raised to the power of the second."*

### Practice: the `Math` class

Here are four real entries you can use on the AP exam:

| Entry | What it does |
|---|---|
| `static int abs(int x)` | Returns the absolute value of `x` |
| `static double abs(double x)` | Returns the absolute value of `x` |
| `static double pow(double base, double exponent)` | Returns `base` raised to the `exponent` |
| `static double sqrt(double x)` | Returns the positive square root of `x` |
| `static double random()` | Returns a `double` that is at least `0.0` and less than `1.0` |

Two things to notice:

1. **`abs` appears twice.** Same name, different parameter types, different return types. Java picks the right one based on what you pass it. That's why `Math.abs(-7)` gave you `7` (an `int`) rather than `7.0`.
2. **`random()` has empty parentheses.** It needs nothing from you, but it's still a behavior — it *does* something — so the parentheses stay.

> The full list of classes and methods you may use on the exam is called the **AP Java Quick Reference**, and it's provided to you during the exam. You don't have to memorize method lists. You *do* have to be able to read them fluently.

````java
public class ReadingTheDocs {
    public static void main(String[] args) {
        // static double sqrt(double x)  -> give it a double, get a double
        double side = Math.sqrt(64);
        System.out.println("A square counter of area 64 has sides of " + side + " ft");

        // static double pow(double base, double exponent)  -> two doubles in, one out
        System.out.println("2 to the 10th power: " + Math.pow(2, 10));

        // static int abs(int x)  -> int in, INT out (note: no .0)
        System.out.println("Register off by: " + Math.abs(237 - 250) + " dollars");

        // static double abs(double x)  -> double in, DOUBLE out (note the .0)
        System.out.println("Same numbers as doubles: " + Math.abs(237.0 - 250.0));
    }
}

ReadingTheDocs.main(null);
````

Compare the last two lines carefully: **same method name, same numbers, and the answers print as `13` and `13.0`.**

Nothing about the values changed — only the *types* of what you passed in, which sent Java to a different documented entry with a different return type. The docs told you that would happen before you ran a single line. That's the whole point of this lesson.

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Library | A collection of pre-written classes available to your program |
| API | Documentation describing what a class provides and how to use it |
| Implementation | The hidden internal code that makes a class actually work |
| Abstraction | Using something through its interface without knowing its internals |
| Attribute | Data a class knows — a noun, no parentheses |
| Behavior | Something a class does — a method, with parentheses |
| Package | A named grouping of related classes, such as `java.util` |
| `import` | A statement making a class from another package available |
| `java.lang` | The package imported automatically into every Java program |
| AP Java Quick Reference | The API subset provided to you during the AP exam |

### The five things that will show up on the exam

1. A **library** is the code; an **API** is the description of how to use it.
2. An API tells you **what**, never **how** — that's abstraction.
3. **Attributes** are nouns without parentheses; **behaviors** are verbs with them.
4. `java.lang` (`String`, `Math`, `System`, `Integer`) needs no import. Everything else does.
5. `cannot find symbol` on a class name means a **missing import**.
