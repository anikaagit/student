## Part 1 — What is an algorithm?

> **Definition:** An **algorithm** is a finite set of precise instructions that can be carried out to solve a problem or accomplish a task.

Break that apart, cafe-style:

| Property | What it means | At the cafe |
|---|---|---|
| **Finite** | It ends. | The drink gets handed over; you don't shake it forever. |
| **Precise / unambiguous** | Every step means exactly one thing. | "Add 30 g of tapioca," not "add some pearls." |
| **Ordered** | Steps happen in a specific sequence. | Seal the cup *after* pouring, not before. |
| **Accomplishes a task** | There's an actual goal. | A finished milk tea. |

Here's the difference between an instruction and an *algorithm*:

- **Not an algorithm:** "Make a milk tea." — a goal, not an algorithm. Way too vague for a computer.
- **An algorithm:** "Boil water. Cook pearls 25 minutes. Steep tea 5 minutes. Combine with 100 mL milk. Seal. Serve." — finite, ordered, precise.

An algorithm is language-independent — you can write one in English, in a flowchart, or in doodles on a napkin. A **program** is what you get when you write an algorithm in a language a computer understands, like Java.

Run the cell below to see an algorithm expressed as a Java program.

````java
public class BobaRecipe {
    public static void main(String[] args) {
        System.out.println("BYTE-SIZED BOBA -- Classic Milk Tea");
        System.out.println("Step 1: Boil 500 mL of water");
        System.out.println("Step 2: Cook tapioca pearls for 25 minutes");
        System.out.println("Step 3: Steep 2 tea bags for 5 minutes");
        System.out.println("Step 4: Add pearls, tea, 100 mL milk, and ice to the cup");
        System.out.println("Step 5: Seal the cup and add a wide straw");
        System.out.println("Order up!");
    }
}

BobaRecipe.main(null);
````

Notice: the computer did the steps **top to bottom, one at a time, exactly once**. That's not an accident — it's the first of three building blocks.

## Part 2 — The three building blocks

Every algorithm you will ever write in this course is built from exactly three control structures. The College Board wants you to be able to name and spot all three.

| Structure | What it does | Cafe version |
|---|---|---|
| **Sequencing** | Statements run in order, one after another | Follow the recipe card top to bottom |
| **Selection** | A condition decides which statements run | "If they ordered jelly, scoop jelly; otherwise scoop pearls" |
| **Iteration** | A block of statements repeats | "Seal every cup in the tray" |

### 2a. Sequencing — order matters

Same five instructions, different order, catastrophically different drink. Run this and picture the mess.

````java
public class OutOfOrder {
    public static void main(String[] args) {
        System.out.println("1. Seal the cup with the plastic film");
        System.out.println("2. Pour in the milk tea");
        System.out.println("3. Drop in the tapioca pearls");
        System.out.println("4. Hand it to the customer");
        System.out.println("...why is there tea all over the counter?");
    }
}

OutOfOrder.main(null);
````

The instructions were all *correct*. The **sequence** was wrong. Computers never say "hey, did you mean to do that?" — they do precisely what you wrote, in the order you wrote it.

### 2b. Selection — the program makes a choice

The cafe doesn't make one drink; it reacts to what the customer wants. That's **selection**: a condition (`true` or `false`) picks which path runs.

Change `topping` to `"jelly"` or `"pudding"` and re-run to take different paths through the same program.

````java
public class ToppingChoice {
    public static void main(String[] args) {
        String topping = "pearls";   // <-- try "jelly" or "pudding", then re-run

        if (topping.equals("pearls")) {
            System.out.println("Scooping warm tapioca pearls...");
        } else if (topping.equals("jelly")) {
            System.out.println("Scooping lychee jelly...");
        } else {
            System.out.println("Sorry, we're out of " + topping + " today.");
        }

        System.out.println("Drink finished.");
    }
}

ToppingChoice.main(null);
````

Only **one** of those three branches ever runs. The last line runs no matter what, because it's outside the `if`.

> You'll go deep on `if` statements in **Unit 2**. For 1.1 you only need to recognize selection when you see it.

### 2c. Iteration — repeat without retyping

Six drinks in the tray, six identical seals. You could copy-paste six `println` statements... or you could loop.

````java
public class SealCups {
    public static void main(String[] args) {
        for (int cup = 1; cup <= 6; cup++) {
            System.out.println("Sealing cup #" + cup);
        }
        System.out.println("Tray ready for pickup.");
    }
}

SealCups.main(null);
````

One instruction, written once, executed six times. Try changing `6` to `20` — notice how little you had to edit. That's the whole point of iteration.

### 2d. All three at once

Real programs mix all three. Read the comments to spot each structure.

> Don't stress about the `String[] orders` array syntax — that's Unit 4. Just read it as "a list of drinks."

````java
public class MorningRush {
    public static void main(String[] args) {
        // SEQUENCING: set up before the rush, in order
        String[] orders = {"taro", "matcha", "taro", "brown sugar"};
        int taroCount = 0;
        System.out.println("Opening the shop...");

        // ITERATION: handle every order in the queue
        for (int i = 0; i < orders.length; i++) {
            String drink = orders[i];

            // SELECTION: taro needs an extra blending step
            if (drink.equals("taro")) {
                taroCount++;
                System.out.println("Order " + (i + 1) + ": taro milk tea (blend the powder first!)");
            } else {
                System.out.println("Order " + (i + 1) + ": " + drink + " milk tea");
            }
        }

        // SEQUENCING again: wrap up
        System.out.println("Taro drinks made this rush: " + taroCount);
    }
}

MorningRush.main(null);
````

## Part 3 — Input and output

> **Program input** is data sent *to* a program for processing.
> **Program output** is any data sent *from* a program to a screen, speaker, file, or another device.

At the cafe, the order slip is the input and the finished drink plus the receipt is the output. Input can come from a keyboard, a mouse click, a tap on a screen, a file, a sensor, or another program.

In the cell below, the four variables at the top act as the input. **Change them and re-run** — different input, different output, same program.

````java
public class OrderSlip {
    public static void main(String[] args) {
        // ---------- INPUT (edit these!) ----------
        String customer      = "Riley";
        String drink         = "Brown Sugar Milk Tea";
        int    quantity      = 2;
        double pricePerDrink = 5.75;

        // ---------- PROCESSING ----------
        double subtotal = quantity * pricePerDrink;
        double tax      = subtotal * 0.0775;
        double total    = subtotal + tax;

        // ---------- OUTPUT ----------
        System.out.println("===== BYTE-SIZED BOBA =====");
        System.out.println("Customer: " + customer);
        System.out.println(quantity + "x " + drink);
        System.out.printf("Subtotal: $%.2f%n", subtotal);
        System.out.printf("Tax:      $%.2f%n", tax);
        System.out.printf("TOTAL:    $%.2f%n", total);
        System.out.println("===========================");
    }
}

OrderSlip.main(null);
````

## Part 4 — How Java actually gets made: the compiler and the JVM

Here's the cafe version. You write an order in your own handwriting. The cashier rewrites it onto a **standardized ticket** using codes every location understands. Then a barista in San Diego, Taipei, or Toronto reads that same ticket and makes your drink — on whatever equipment their shop happens to have.

That's Java in three steps:

```
   BobaRecipe.java          BobaRecipe.class            any computer
 ┌──────────────────┐      ┌──────────────────┐       ┌──────────────────┐
 │   SOURCE CODE    │      │     BYTECODE     │       │     OUTPUT       │
 │  human-readable  │─────►│  the standard    │──────►│  program runs    │
 │  Java you type   │javac │  "order ticket"  │  JVM  │  Mac / PC / phone│
 └──────────────────┘      └──────────────────┘       └──────────────────┘
      you write it          the COMPILER makes it       the JVM runs it
```

| Term | What it is |
|---|---|
| **Source code** | The Java you write, saved in a `.java` file. Readable by humans. |
| **Compiler** (`javac`) | A program that translates your source code into bytecode. It also checks your syntax and refuses to translate broken code. |
| **Bytecode** | The `.class` file. Not English, not machine code — an in-between language that every JVM understands. |
| **JVM** (Java Virtual Machine) | Software installed on a device that reads bytecode and executes it using that device's actual hardware. |

### Why anyone cares

Because the JVM does the device-specific part, **one** compiled `.class` file runs on Windows, macOS, Linux, and Android without being rewritten. That's the slogan you'll hear: *write once, run anywhere*. One recipe card, every franchise location.

> **Exam-worthy detail:** the compiler runs *before* your program does. If the compiler finds a problem, your program never runs at all — not even the first line.

## Part 5 — Anatomy of a Java program

Every line below is doing a specific job. Read the comments, then run it.

````java
/**
 * HelloBoba -- a Javadoc comment (starts with slash-star-star).
 * Used to describe a whole class or method.
 */
public class HelloBoba {                       // class name matches the file name: HelloBoba.java

    public static void main(String[] args) {   // main() is where every Java program starts

        System.out.println("Welcome to Byte-Sized Boba!");  // one statement; ends with a semicolon

        /* A block comment can
           span multiple lines. */

        System.out.print("Order up: ");         // print() does NOT move to a new line
        System.out.println("Taro Milk Tea");    // println() DOES move to a new line

        // A single-line comment. The compiler ignores every comment on this page.
    }
}

HelloBoba.main(null);
````

### The rules hiding in that code

| Rule | Why it matters |
|---|---|
| Code lives inside a **class** | `public class HelloBoba { ... }` wraps everything. |
| Execution starts at **`main`** | `public static void main(String[] args)` is the front door. |
| Statements end with a **semicolon** `;` | Missing one is the #1 beginner compile error. |
| Blocks are wrapped in **braces** `{ }` | Every `{` needs a matching `}`. |
| Java is **case sensitive** | `System` ≠ `system`, `myDrink` ≠ `MyDrink`. |
| **Comments** are ignored by the compiler | `//`, `/* */`, and `/** */` are for humans only. |

Case sensitivity trips people up constantly, so let's prove it.

````java
public class CaseMatters {
    public static void main(String[] args) {
        String cupSize = "large";
        String CupSize = "small";   // legal Java, but a COMPLETELY separate variable

        System.out.println("cupSize is " + cupSize);
        System.out.println("CupSize is " + CupSize);

        // Uncomment the next line and re-run to watch the compiler reject it:
        // System.out.PRINTLN("this will not compile");
    }
}

CaseMatters.main(null);
````

## Part 6 — Three ways to break a drink (and a program)

This is the highest-value part of 1.1. AP loves asking you to classify an error.

| Error type | When it happens | Cafe analogy |
|---|---|---|
| **Compile-time (syntax) error** | Before the program runs. The compiler can't understand your code, so nothing executes. | The order is written in handwriting nobody can read. The drink never gets started. |
| **Run-time error** | While the program is running. It starts fine, then crashes partway. | The machine runs out of tapioca halfway through. Half a drink, then everything stops. |
| **Logic error** | Never crashes. The program runs perfectly and gives the *wrong answer*. | You grabbed salt instead of sugar. It looks like a milk tea. It gets served. It's wrong. |

Logic errors are the dangerous ones — the computer will never warn you, because from its point of view nothing went wrong.

### 6a. Compile-time (syntax) error

**Heads up: the next cell is supposed to fail.** Run it, then read the error message carefully — learning to read compiler errors is a real skill.

````java
public class BrokenOrder {
    public static void main(String[] args) {
        System.out.println("Making a taro milk tea")   // BUG 1: what's missing at the end?
        system.out.println("Adding pearls...");        // BUG 2: look very closely at "system"
    }
}

BrokenOrder.main(null);
````

Two bugs in two lines: a missing semicolon, and `system` that should be `System` (case sensitivity again).

Two things to notice:

- **Nothing printed at all** — not even the first line. The compiler stopped everything before execution began.
- The error message probably only mentioned the **first** bug. Compilers often give up at the first thing they can't parse, so fixing one error and re-running to reveal the next is completely normal. Don't panic when the error count goes up before it goes down.

Here's the fixed version:

````java
public class FixedOrder {
    public static void main(String[] args) {
        System.out.println("Making a taro milk tea");
        System.out.println("Adding pearls...");
    }
}

FixedOrder.main(null);
````

### 6b. Run-time error

This one compiles just fine. Run it and watch where it dies.

````java
public class SplitTheBill {
    public static void main(String[] args) {
        int totalCents = 1725;
        int friends = 0;              // ...nobody else showed up

        System.out.println("Splitting the bill...");
        int each = totalCents / friends;   // dividing by zero: crash happens right here
        System.out.println("Each person owes " + each + " cents");
    }
}

SplitTheBill.main(null);
````

Look at the difference from 6a: `"Splitting the bill..."` **did** print. The program compiled, started, ran a couple of statements, and *then* threw an `ArithmeticException`. The line after the crash never ran.

That's the fingerprint of a run-time error: partial output, then a crash.

### 6c. Logic error

No red text. No crash. Just a wrong answer.

Byte-Sized Boba gives **10 loyalty points per drink**. A customer buys 4 drinks, so they should get 40 points. Run the cell.

````java
public class LoyaltyPoints {
    public static void main(String[] args) {
        int drinksBought   = 4;
        int pointsPerDrink = 10;

        int points = drinksBought + pointsPerDrink;   // hmm.

        System.out.println("Drinks bought: " + drinksBought);
        System.out.println("Points earned: " + points);
        System.out.println("Thanks for visiting Byte-Sized Boba!");
    }
}

LoyaltyPoints.main(null);
````

It printed `14`. The customer should have gotten `40`.

Java did **exactly** what it was told — `4 + 10` really is `14`. The instruction was valid; the *thinking behind it* was wrong. That's a logic error, and the only thing that catches it is you testing your program against an answer you worked out yourself.

<details>
<summary><b>Fix (click to reveal)</b></summary>

Change the operator from `+` to `*`:

```java
int points = drinksBought * pointsPerDrink;   // 4 * 10 = 40
```
</details>

**1.** Which of the following best describes an algorithm?

&nbsp;&nbsp;(A) A programming language used to write software
&nbsp;&nbsp;(B) A finite set of precise instructions for accomplishing a task

**2.** What does the Java compiler produce, and what executes it?

&nbsp;&nbsp;(A) Source code, executed by the programmer
&nbsp;&nbsp;(B) Machine code specific to one processor, executed directly by the CPU
&nbsp;&nbsp;(C) Bytecode, executed by the Java Virtual Machine
&nbsp;&nbsp;(D) An algorithm, executed by the compiler

<details><summary>Answer</summary>

**(C)**. `javac` turns source code into bytecode (`.class`), and the JVM executes that bytecode. Because the JVM handles the device-specific part, the same bytecode runs on many platforms.
</details>

**3.** A program compiles and runs with no error messages, but the total on every receipt is exactly double what it should be. What kind of error is this?

&nbsp;&nbsp;(A) Compile-time error

&nbsp;&nbsp;(C) Logic error
