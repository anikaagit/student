## Part 1 — The compiler ignores every word

Before anything else, prove it to yourself.

Comments are stripped out before your code is compiled. You can write complete nonsense inside one and the program still runs perfectly. Comments never execute, never affect output, and never slow anything down.

````java
public class CommentsIgnored {
    public static void main(String[] args) {
        // This line does absolutely nothing.

        /* Neither does this one, even though it contains
           total gibberish:  %%%$$$ int int int ))));;;;
           The compiler never even looks at it. */

        System.out.println("Only this line produced output.");

        // System.out.println("This statement is commented out, so it never runs.");
    }
}

CommentsIgnored.main(null);
````

That last line shows a second use for comments: **temporarily disabling code** while you test something. You've already done this a few times in earlier lessons.

## Part 2 — The three comment types

| Type | Syntax | Use it for |
|---|---|---|
| **Single-line** | `// note` | A quick note about one line or a short section |
| **Block** | `/* note */` | Explanations spanning several lines; disabling chunks of code |
| **Javadoc** | `/** note */` | Describing a class or method — **generates API documentation** |

All three are ignored at run time. The difference is who reads them and how.

That third one is the important one, and here's why it's different in kind rather than just in style:

> **Javadoc comments are read by a tool.** The `javadoc` program that ships with Java scans `/** */` comments and produces browsable HTML documentation automatically.

This is the direct answer to something from Lesson 1.7. When you read the documentation for `Math.sqrt`, you were reading **generated Javadoc**. Somebody at Oracle wrote a `/** */` comment above the method, and a tool turned it into the page you read. The API you consumed last lesson and the comments you write today are the same thing at two different stages.

A Javadoc comment goes **directly above** whatever it describes — a class, a method, or a constant.

````java
/**
 * DrinkPricer calculates order totals for Byte-Sized Boba.
 * A Javadoc comment above a class describes the whole class.
 */
public class DrinkPricer {

    /** The sales tax rate charged at all locations. */
    public static final double TAX_RATE = 0.0775;

    public static void main(String[] args) {
        // Single-line comment: a quick note
        double subtotal = 11.50;

        /* Block comment: use this when the explanation
           genuinely needs more than one line to make sense. */
        double total = subtotal + subtotal * TAX_RATE;

        System.out.println("Subtotal: $" + subtotal);
        System.out.println("Total:    $" + total);
    }
}

DrinkPricer.main(null);
````

> **One syntax gotcha:** block comments **do not nest**. If you write `/*` inside another `/* */`, the first `*/` ends the whole thing and the leftovers become code that won't compile. This bites people who try to comment out a section that already contains a block comment.

## Part 3 — Comments worth reading

Writing a comment is easy. Writing a *useful* one takes judgment.

The most common beginner mistake is a comment that simply restates the code in English:

```java
cups--;             // decrease cups by 1
```

Anyone reading that already knows `--` subtracts one. The comment added nothing, and now there are **two** things to keep updated instead of one. If the code changes and the comment doesn't, the comment becomes an active lie.

> **The rule: code says WHAT. Comments say WHY.**

Compare these two programs. They are **identical code** with identical output.

````java
public class BadComments {
    public static void main(String[] args) {
        int cups = 40;                    // set cups to 40
        cups--;                           // decrease cups by 1
        cups /= 2;                        // divide cups by 2
        System.out.println("Cups: " + cups);   // print cups
    }
}

BadComments.main(null);
````

````java
public class GoodComments {
    public static void main(String[] args) {
        int cups = 40;    // morning delivery arrives as one 40-cup case
        cups--;           // one cup is always used to calibrate the sealing machine
        cups /= 2;        // stock is split evenly between counter and drive-thru

        System.out.println("Cups for the front counter: " + cups);
    }
}

GoodComments.main(null);
````

Same instructions, same `19`. But only the second version tells you that the sealer eats a cup, or that the halving is a policy decision rather than a math trick. **The first program's comments could be deleted with zero information lost. The second program's could not.**

### A checklist for good comments

Comment when you're explaining:

- **Why** a decision was made — `// tax is added before the discount per state law`
- A **non-obvious** consequence — `// integer division: partial dollars earn no points`
- A **unit or format** the code can't show — `// temperature in degrees Fahrenheit`
- A **precondition or postcondition** (Parts 5 and 6)

Don't comment:

- Anything the code already says plainly
- Something you can fix with a better variable name instead

That last one deserves emphasis. `int d = 7; // days until restock` is worse than `int daysUntilRestock = 7;` with no comment at all. **A good name beats a good comment**, because a name can't fall out of sync with the code.

## Part 4 — Javadoc tags

Javadoc comments use structured **tags** so the generated documentation can be organized consistently.

| Tag | Describes |
|---|---|
| `@param name` | One input the method needs, and any restrictions on it |
| `@return` | What the method gives back |

A complete method Javadoc has a short description first, then one `@param` per parameter, then `@return`.

Read the comment below and notice you could use `calculateTotal` correctly **without reading a single line of its body** — which is exactly the test of good documentation.

````java
public class PriceCalculator {

    /**
     * Calculates the total cost of an order, including sales tax.
     *
     * @param quantity the number of drinks ordered
     * @param unitPrice the price of one drink, in dollars
     * @return the total cost in dollars with 7.75% tax applied
     */
    public static double calculateTotal(int quantity, double unitPrice) {
        double subtotal = quantity * unitPrice;
        return subtotal + subtotal * 0.0775;
    }

    public static void main(String[] args) {
        System.out.println("2 drinks at $5.75: $" + calculateTotal(2, 5.75));
        System.out.println("3 drinks at $6.50: $" + calculateTotal(3, 6.50));
    }
}

PriceCalculator.main(null);
````

Notice what the documentation had to specify that the method name alone could never tell you: **the total already includes tax.** Without that sentence, a caller might reasonably add tax a second time.

## Part 5 — Preconditions

> **Definition:** A **precondition** is a condition that must be true *before* a method runs in order for the method to behave as expected.

The cafe version: the sealing machine's instructions say *"the machine must be switched on and a cup must be in the cradle."* Those are preconditions. Press the button with an empty cradle and you get a mess — and that's not the machine's fault.

Here's the part the AP exam tests, and it surprises almost everyone:

> **The method is not required to check whether the precondition is true. Satisfying it is the caller's responsibility.**

A precondition is a **contract**. It says: *"if you give me valid input, I promise to work. If you don't, all bets are off — I might crash, or I might just hand you garbage."*

Read the method below, then look at what happens when the contract is broken.

````java
public class BillSplitter {

    /**
     * Splits an order total evenly among a group of people.
     *
     * Precondition: people > 0
     *
     * @param total the order total in dollars
     * @param people the number of people splitting the bill
     * @return each person's share, in dollars
     */
    public static double splitBill(double total, int people) {
        return total / people;      // no check for zero -- see the precondition
    }

    public static void main(String[] args) {
        System.out.println("$24.00 among 4: $" + splitBill(24.00, 4));
        System.out.println("$24.00 among 3: $" + splitBill(24.00, 3));

        // Precondition VIOLATED. The method does not protect you.
        System.out.println("$24.00 among 0: $" + splitBill(24.00, 0));
    }
}

BillSplitter.main(null);
````

Look at that last line: **`Infinity`**. Not an error message. Not a crash. A value that is mathematically meaningless as a per-person share, printed onto a receipt as though it were a dollar amount.

That's the roundoff-and-range material from 1.5 doing exactly what it warned about: `double` division by zero produces `Infinity` rather than throwing. Had `total` been an `int`, this same call would have thrown an `ArithmeticException` instead — a loud failure rather than a silent one.

Either way, **the method behaved correctly.** It never promised anything for `people == 0`. The caller broke the contract.

> **On the AP exam:** if documentation states a precondition, assume it is satisfied. Do not write code to validate it unless the question explicitly asks you to.

Preconditions usually restrict:

- **Ranges** — `score >= 0 && score <= 100`
- **Non-zero divisors** — `people > 0`
- **Non-null references** — `name is not null`
- **Relationships between parameters** — `startIndex <= endIndex`

## Part 6 — Postconditions

> **Definition:** A **postcondition** is a condition that is true *after* the method finishes. It's what the method **promises**.

If a precondition is what the method demands, a postcondition is what it guarantees in return. Together they form the two halves of the contract:

```
   ┌────────────────────────────────────────────────┐
   │  PRECONDITION          POSTCONDITION           │
   │  what the CALLER       what the METHOD         │
   │  must guarantee   ──►  guarantees back         │
   │                                                │
   │  "people > 0"          "returns each share"    │
   └────────────────────────────────────────────────┘
```

Postconditions typically describe the **return value**, any **change to an object's state**, or a guaranteed property of the result.

The method below has a postcondition that captures something genuinely non-obvious — and it's the kind of detail that causes bugs when it goes undocumented.

````java
public class TrayPacker {

    /**
     * Counts how many completely full trays a batch of cups will fill.
     *
     * Precondition:  cups >= 0 and cupsPerTray > 0
     * Postcondition: returns a value >= 0; leftover cups that do not
     *                fill a complete tray are NOT counted
     *
     * @param cups the total number of cups to pack
     * @param cupsPerTray how many cups one tray holds
     * @return the number of completely filled trays
     */
    public static int fullTrays(int cups, int cupsPerTray) {
        return cups / cupsPerTray;
    }

    public static void main(String[] args) {
        System.out.println("20 cups, 6 per tray -> " + fullTrays(20, 6) + " full trays");
        System.out.println(" 5 cups, 6 per tray -> " + fullTrays(5, 6) + " full trays");
        System.out.println("12 cups, 6 per tray -> " + fullTrays(12, 6) + " full trays");
    }
}

TrayPacker.main(null);
````

Five cups returned **zero** trays. Without that postcondition spelled out, a caller might reasonably expect `1` (a partly filled tray) and ship five cups to a customer who ordered a full one.

The postcondition documents a consequence of integer division that the method name `fullTrays` only hints at. **This is the single most valuable thing documentation does: state the surprising part out loud.**

**1.** Which comment type is used to generate API documentation?

&nbsp;&nbsp;(A) `// comment` &nbsp;&nbsp; (B) `/* comment */` &nbsp;&nbsp; (C) `/** comment */` &nbsp;&nbsp; (D) `# comment`

**2.** What effect do comments have on a running program?

&nbsp;&nbsp;(A) They print to the console when reached
&nbsp;&nbsp;(B) They slow execution slightly
&nbsp;&nbsp;(C) They have no effect; the compiler ignores them
&nbsp;&nbsp;(D) They are stored in memory as Strings

<details><summary>Answer</summary>

**(C)**. Comments are stripped before compilation. They never execute, produce output, or occupy memory at run time.

**3.** A method is documented with `Precondition: index >= 0`. A programmer calls it with `index = -5`. Which statement is correct?

&nbsp;&nbsp;(A) The method must throw an exception

&nbsp;&nbsp;(C) The method's behavior is not guaranteed; the caller violated the contract

**(C)**. A precondition places the responsibility on the **caller**. The method is not required to check it, and when it's violated the method makes no promises at all — it might crash, or it might silently return nonsense.

# Closing time

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Comment | Text ignored by the compiler, written for human readers |
| Single-line comment | `// text` — runs to the end of the line |
| Block comment | `/* text */` — spans multiple lines; does not nest |
| Javadoc comment | `/** text */` — used to generate API documentation |
| `@param` | Javadoc tag describing one method parameter |
| `@return` | Javadoc tag describing what a method gives back |
| Precondition | What must be true **before** a method runs; the **caller's** responsibility |
| Postcondition | What is guaranteed true **after** a method finishes; the method's promise |

### The five things that will show up on the exam

1. Comments are **ignored by the compiler** — no output, no effect, no cost.
2. `/** */` is Javadoc and **generates API documentation**; `//` and `/* */` do not.
3. A **precondition** is what the caller must guarantee.
4. **A method is not required to check its preconditions.** Assume they hold.
5. A **postcondition** is what the method promises — usually about the return value.

### The one-sentence version

> **Code says what. Comments say why. Preconditions say what you owe the method. Postconditions say what it owes you.**
