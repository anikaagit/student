## You've been building objects the whole time

Here's an uncomfortable question: in five words or fewer, what's the difference between a `String` and an `int`?

You could probably answer *"one's a reference type"* — that's been sitting in your vocabulary since Lesson 1.2. But *why* is `String` a reference type, and what does "reference" actually refer to? What is `String` even a reference **to**?

That's today's lesson. The cafe's recipe card for Taro Milk Tea is not a drink. It's a **blueprint** — instructions for making a drink. That relationship — one blueprint, many independent things built from it — is exactly the relationship between a **class** and an **object**.

## Part 1 — Class as blueprint, object as instance

> **A class is a blueprint.** It describes what something is made of and what it can do, without being an actual "something" itself.
> **An object is an instance of a class** — one concrete thing built following that blueprint.

Nobody drinks a recipe card. Nobody drinks `String` either. `String` describes *what a string of text is and does* — that it holds characters, that it has a `length()`, that it can be compared. Every time you write `"Taro Milk Tea"`, Java builds an actual `String` **object**: one concrete, particular piece of text living in memory, built according to the `String` blueprint.

```
             CLASS (the blueprint)                 OBJECT (an instance)
    ┌───────────────────────────────┐      ┌──────────────────────────┐
    │  String                       │      │  "Taro Milk Tea"         │
    │  -- describes: holds text     │─────►│  one particular String,  │
    │  -- describes: has length()   │      │  built from the blueprint│
    └───────────────────────────────┘      └──────────────────────────┘
```

You don't only build one object per class, either. Just like one recipe card produces every cup of taro milk tea the cafe ever serves, one `String` class produces every piece of text every Java program on Earth has ever created.

````java
public class ClassVsObject {
    public static void main(String[] args) {
        // Three SEPARATE String objects, all built from the SAME String class/blueprint.
        String drink1 = "Taro Milk Tea";
        String drink2 = "Matcha Latte";
        String drink3 = "Thai Tea";

        System.out.println(drink1);
        System.out.println(drink2);
        System.out.println(drink3);

        System.out.println();
        System.out.println("Three different objects. One shared blueprint: the String class.");
    }
}

ClassVsObject.main(null);
````

## Part 2 — Objects are independent

This is the property that actually makes objects useful: **two objects built from the same class don't affect each other.** Changing one never touches the other, even though they were stamped from identical instructions.

`Scanner` makes this easy to see, because a `Scanner` object keeps track of **where it currently is** in its input — and that position belongs to one specific `Scanner`, not to the `Scanner` class in general.

````java
import java.util.Scanner;

public class IndependentObjects {
    public static void main(String[] args) {
        // Two SEPARATE Scanner objects, each with its OWN position in its OWN input.
        Scanner scanA = new Scanner("Taro Milk Tea");
        Scanner scanB = new Scanner("Matcha Latte");

        System.out.println("scanA reads: " + scanA.next());   // "Taro"
        System.out.println("scanB reads: " + scanB.next());   // "Matcha"

        // Advancing scanA does NOT move scanB. They are independent objects.
        System.out.println("scanA reads next: " + scanA.next());          // "Milk"
        System.out.println("scanB is unaffected, still reads: " + scanB.next());   // "Latte"
    }
}

IndependentObjects.main(null);
````

`scanA` reading further into its own input had zero effect on `scanB`. Both objects came from the identical `Scanner` blueprint, and yet each one tracks its own separate position — because each object gets its **own copy** of whatever data the class describes. This is the recipe-card idea made concrete: one card, unlimited independent cups, none of them sharing tea.

## Part 3 — Declaring a reference-type variable

Back in Lesson 1.2, you saw this exact picture:

```
  PRIMITIVE                          REFERENCE
  int cupsSold = 12;                 String name = "Riley";

  cupsSold                           name
 ┌──────────┐                       ┌──────────┐        ┌───────────┐
 │    12    │  <- the value          │  0x4a2f  │ ─────► │  "Riley"  │
 └──────────┘     lives right here   └──────────┘        └───────────┘
                                      an address          the actual
                                      to somewhere         object
```

Now you can say precisely what that arrow means: **a reference-type variable doesn't hold the object itself — it holds a reference (an address) pointing to where the object actually lives.** The variable `name` is not the text `"Riley"`. It's a pointer that says "the actual object is over there."

The **declaration syntax is identical** to declaring a primitive from Lesson 1.2 — `Type variableName;` — the difference is entirely in what gets stored once you assign something.

````java
import java.util.Scanner;

public class DeclaringReferences {
    public static void main(String[] args) {
        // Declaration syntax: Type variableName; -- same shape as a primitive from 1.2
        String customerName;
        Scanner orderScanner;

        // Assignment gives each variable a REFERENCE to a real object
        customerName = "Riley";
        orderScanner = new Scanner("2 Taro Milk Tea");

        System.out.println("customerName refers to: " + customerName);
        System.out.println("orderScanner's first token: " + orderScanner.next());
    }
}

DeclaringReferences.main(null);
````

That worked exactly like declaring an `int` and assigning it later (Lesson 1.4) — declare first, assign after, on separate lines. The only thing that changed is *what kind of thing* gets stored: an `int` variable stores a value directly; `customerName` and `orderScanner` store references to objects sitting elsewhere.

### Reading before assigning still fails, for the same reason as 1.4

**Heads up: the next cell is supposed to fail.**

````java
public class TooSoonReference {
    public static void main(String[] args) {
        String flavor;                          // declared, but no object assigned yet
        System.out.println("Flavor: " + flavor); // trying to read an empty reference
    }
}

TooSoonReference.main(null);
````

`variable flavor might not have been initialized` — the exact same compile-time protection from Lesson 1.4, now applied to a reference type instead of a primitive. Java refuses to let you read a variable that has never been given anything to refer to.

## Part 4 — `null`: a reference to nothing

Primitives have no way to say "I have no value" — an `int` is always some number. Reference types are different: they can deliberately point at **nothing at all**.

> **`null` is a literal meaning "this reference variable does not currently point to any object."**

This is different from the uninitialized error you just saw. Assigning `null` is a real, legal assignment — it just assigns "nothing" on purpose, instead of leaving the variable empty and unassigned.

````java
import java.util.Scanner;

public class NullReferences {
    public static void main(String[] args) {
        String flavor = null;      // explicitly refers to NO object -- this is legal
        Scanner backup = null;     // same idea for any reference type

        System.out.println("flavor is: " + flavor);
        System.out.println("backup is: " + backup);

        flavor = "Taro";           // now it refers to a real object
        System.out.println("flavor is now: " + flavor);
    }
}

NullReferences.main(null);
````

Printing a `null` reference doesn't crash — it prints the word `null`, harmlessly. That's worth remembering, because it's about to matter: `null` is genuinely useful for representing "no drink selected yet" or "no customer assigned to this order," in a way a primitive simply cannot express.

> **A preview, not today's problem:** *calling a method* on a `null` reference — like `flavor.length()` when `flavor` is `null` — is a different story entirely, and it crashes. That's `NullPointerException`, and it's the formal subject of Topic **1.14**. Today, just know that `null` itself is a legal, harmless value to store and print.

### The full picture: declared, uninitialized, null, and assigned

| State | Example | Can you print it? |
|---|---|---|
| Declared, never assigned | `String s;` | **No** — compile-time error |
| Assigned `null` | `String s = null;` | **Yes** — prints `null` |
| Assigned a real object | `String s = "Taro";` | **Yes** — prints `Taro` |

## Part 5 — Putting it together: the order queue

One more example, using everything from this lesson. Three separate `Scanner` objects simulate three separate customers' orders arriving — same class, same code building each one, completely independent results.

````java
import java.util.Scanner;

public class OrderQueue {
    public static void main(String[] args) {
        // Three objects, one class. Each is a completely independent order.
        Scanner order1 = new Scanner("Riley 2 TaroMilkTea");
        Scanner order2 = new Scanner("Jordan 1 MatchaLatte");
        Scanner order3 = new Scanner("Sam 3 ThaiTea");

        printOrder(order1);
        printOrder(order2);
        printOrder(order3);
    }

    public static void printOrder(Scanner order) {
        String name     = order.next();
        int    quantity = order.nextInt();
        String drink    = order.next();
        System.out.println(name + " ordered " + quantity + "x " + drink);
    }
}

OrderQueue.main(null);
````

`printOrder` is a single method that gets handed a **different object** each time it's called (this is a method parameter of a reference type — the same idea as any other parameter from Lesson 1.9, just holding a reference this time instead of a primitive). Nothing about `order1` leaked into `order2` or `order3`. Three blueprints, three independent objects, three independent results — the entire lesson in one program.

# Closing time

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Class | A blueprint describing what objects of that type know and can do |
| Object | A specific instance built according to a class's blueprint |
| Instance | Another word for a specific object built from a class |
| Reference-type variable | A variable that stores an address pointing to an object, not the object itself |
| `null` | A literal meaning "this reference points to no object" |

### The five things that will show up on the exam

1. A **class** is a blueprint; an **object** is one instance built from it.
2. Objects from the **same class** are still **independent** of each other.
3. A reference-type variable declares with the same syntax as a primitive: `Type name;`.
4. Reading a reference variable before it's assigned is a **compile-time error** — same rule as 1.4, for primitives.
5. `null` is a **legal, explicit value** — different from being uninitialized, and safe to print.
